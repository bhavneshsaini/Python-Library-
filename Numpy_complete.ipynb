## 1.0 Scalar, Vector, Matrix and Tensor — Understand This First

Before creating different NumPy arrays, understand how the **number of dimensions** changes the name we commonly use.

| Concept | NumPy dimensions | Example | Typical shape | Meaning |
|---|---:|---|---|---|
| **Scalar** | 0D | `np.array(7)` | `()` | A single value, such as one age, one price, or one temperature. |
| **Vector** | 1D | `np.array([10, 20, 30])` | `(3,)` | A single ordered list of values. It has one axis. |

| **Tensor** | 3D or higher in this course | `np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])` | `(2, 2, 2)` | A multi-dimensional collection, such as a stack of matrices. |

### Easy way to remember

- **Scalar** → one number
- **Vector** → one line/list of numbers
- **Matrix** → rows × columns
- **Tensor** → multiple dimensions / a stack of matrices

> **Important:** In mathematics and machine learning, *tensor* is the general term and can include 0D scalars, 1D vectors, and 2D matrices. For beginner NumPy ., we will use **tensor** mainly for arrays with **3 or more dimensions**.

## 1.7 `np.arange()` — Create Values Using a Step Size

`np.arange(start, stop, step)` works similarly to Python's `range()` but returns a NumPy array.

- `start` → first value
- `stop` → stopping boundary; it is **not included**
- `step` → gap between consecutive values

Example: `np.arange(1, 10, 2)` gives `1, 3, 5, 7, 9`.

**Remember:** `arange` asks for the **step size**, while `linspace` asks for the **number of values**.


```python
# ============================================================================
# . NOTES
# - .: 1.7 `np.arange()` — Create Values Using a Step Size.
# - np.arange(start, stop, step) creates regularly spaced values using a STEP SIZE; the stop
#   value is excluded.
# ============================================================================

print(np.arange(0, 10))  # Prints the result: creates values from 0 to 9 with the default step of 1; the stop value 10 is excluded.
print(np.arange(0, 100, 10))  # Prints the result: creates values 0, 10, 20, ..., 90 using a step of 10; the stop value 100 is excluded.
print(np.arange(100, 10, -10))  # Prints the result: creates a descending sequence 100, 90, ..., 20; step -10 moves backward and the stop value 10 is excluded.
print(np.arange(1, 10, 2))  # Prints the result: creates 1, 3, 5, 7, 9 using a step of 2; the stop value 10 is excluded.
```

## 1.8 `np.linspace()` — Create Evenly Spaced Values

`linspace` means **linear space**. It creates a fixed **number of evenly spaced values** between a starting value and an ending value.

**Syntax:** `np.linspace(start, stop, num)`

- `start` → first value
- `stop` → last value; it is included by default
- `num` → how many values you want in total
- NumPy calculates the gap automatically.

For example, `np.linspace(10, 20, 4)` creates **4 values** from 10 through 20.  
This is different from `np.arange()`, where the third argument represents the **step size**.


```python
# ============================================================================
# . NOTES
# - .: 1.8 `np.linspace()` — Create Evenly Spaced Values.
# - np.linspace(start, stop, num) creates a chosen NUMBER of evenly spaced values; the stop
#   value is included by default.
# ============================================================================

print(np.linspace(10, 20, 4))  # Prints the result: creates 4 evenly spaced values from 10 to 20; unlike arange, the stop value 20 is included by default.
print(np.linspace(10, 20, 15))  # Prints the result: creates 11 evenly spaced values from 10 to 20, so the spacing is 1; the stop value 20 is included.
```

## 1.14 Reproducible Random Numbers with `seed` — What Does Reproducible Mean?

NumPy random functions generate **pseudo-random** numbers. The values look random, but they are produced by an algorithm.

`np.random.seed(123)` gives that algorithm a fixed **starting point**.

Why use a seed?

- If you run the same code again with the **same seed**, you get the **same sequence of random numbers**.
- This is called **reproducible** or **repeatable** output.
- It is useful in ., testing, debugging, data science experiments, and machine learning because everyone can reproduce the same example.

The number `123` is not special. You can use another integer. What matters is that using the same seed again resets the generator to the same starting point.

> The seed does **not** mean every call always returns one identical number. It makes the **sequence** repeat when the seed is reset.


```python
# ============================================================================
# . NOTES
# - What this cell teac1.14 Reproducible Random Numbers with `seed` — What Does
#   Reproducible Mean?.
# - np.random.randint(low, high, size) generates pseudo-random integers; high is excluded.
# - np.random.seed(number) resets NumPy's pseudo-random generator so the same seed produces the
#   same sequence again—useful for repeatable . examples and tests.
# - np.array_equal(a, b) requires exact equality of shape and values.
# ============================================================================

np.random.seed(123)  # Sets the random seed so the pseudo-random sequence can be reproduced exactly when the same seed is used again.
first = np.random.randint(0, 100, 10)  # Generates random integers in the requested range; the upper limit is excluded.

np.random.seed(123)  # Sets the random seed so the pseudo-random sequence can be reproduced exactly when the same seed is used again.
second = np.random.randint(0, 100, 10)  # Generates random integers in the requested range; the upper limit is excluded.

print(first)  # Prints the current value or result so it can be viewed in the notebook output.
print(second)  # Prints the current value or result so it can be viewed in the notebook output.
print("Same output:", np.array_equal(first, second))  # Prints the result: checks whether two arrays have exactly the same shape and values.
```

## 6.3 Flatten vs Ravel

```python
# ============================================================================
# . NOTES
# - .: 6.3 Flatten vs Ravel.
# - np.array(...) converts a Python sequence (such as a list) into a NumPy ndarray.
# - .flatten() converts an array to 1D and returns a new independent COPY.
# - .ravel() converts an array to 1D and returns a VIEW when possible, so it can be more memory-
#   efficient.
# ============================================================================

"""flatten() → COPY → creates a new array with separate memory
             → memory is NOT shared with the original array (A)
             → changes in the copied array do NOT affect A


ravel()   → VIEW → creates a new 1-D array object
             → memory is SHARED with the original array (A)
             → changes in the view CAN affect A"""

A = np.array([  # Creates a NumPy array from the supplied Python values.
    [1, 2, 3, 4],  # Adds one row/list of values to the array being defined.
    [5, 6, 7, 8],  # Adds one row/list of values to the array being defined.
    [10, 11, 12, 13],  # Adds one row/list of values to the array being defined.
    [14, 15, 16, 17]  # Adds the final row/list of values for the array being defined.
])  # Closes the multiline array, list, or function call started above.

flat_copy = A.flatten()  # Converts a multidimensional array to 1D and returns an independent copy.
ravel_view = A.ravel()  # Converts the array to 1D and usually returns a view that can share memory with the original.

print("flatten():", flat_copy)  # Displays the value/result labeled "flatten()" so it is easy to identify in the output.
print("ravel()  :", ravel_view)  # Displays the value/result labeled "ravel()" so it is easy to identify in the output.
# how to check memory share ? 
print(np.shares_memory(A, flat_copy))    # False
print(np.shares_memory(A, ravel_view))  # True
```

### . Difference
- `flatten()` returns a **copy**.
- `ravel()` returns a **view when possible**, so it is usually more memory-efficient.

## 9.1 Sort a 1D Array

```python
# ============================================================================
# . NOTES
# - .: 9.1 Sort a 1D Array.
# - np.array(...) converts a Python sequence (such as a list) into a NumPy ndarray.
# - np.sort(...) returns a sorted COPY; the original array is unchanged unless the result is
#   assigned back.
# - The slice [::-1] uses a step of -1, which reads elements in reverse order.
# ============================================================================

a = np.array([10, 5, 2, 22, 12, 92, 17, 33])  # Creates a NumPy array from the supplied Python values.

print("Ascending :", np.sort(a))  # Prints the result: returns a sorted copy of the array; axis controls the direction for multidimensional arrays.
print("Descending:", np.sort(a)[::-1])  # Prints the result: returns a sorted copy of the array; axis controls the direction for multidimensional arrays.
```

## 9.3 Shuffle an Array

```python
# ============================================================================
# . NOTES
# - .: 9.3 Shuffle an Array.
# - np.arange(start, stop, step) creates regularly spaced values using a STEP SIZE; the stop
#   value is excluded.
# - np.random.seed(number) resets NumPy's pseudo-random generator so the same seed produces the
#   same sequence again—useful for repeatable . examples and tests.
# - np.random.shuffle(array) rearranges the array in place, so the original array itself is
#   changed.
# ============================================================================

p = np.arange(0, 20)  # Creates regularly spaced values using a step size; the stop value is excluded.
np.random.seed(7)  # Sets the random seed so the pseudo-random sequence can be reproduced exactly when the same seed is used again.
np.random.shuffle(p)  # Shuffles the array values in place, changing their order.

print(p)  # Prints the current value or result so it can be viewed in the notebook output.
```

## 9.4 Find N Largest and N Smallest Values

```python
# ============================================================================
# . NOTES
# - .: 9.4 Find N Largest and N Smallest Values.
# - np.argsort(array) returns the indexes that would place the array in sorted order.
# - np.argpartition(...) partially orders an array and is efficient when you only need the
#   smallest/largest n elements.
# - The slice [::-1] uses a step of -1, which reads elements in reverse order.
# ============================================================================

n = 4  # Stores this scalar (single numeric value) in the variable for use in the next calculation.

largest = p[np.argsort(p)[-n:]][::-1]  # Returns the indexes that would arrange the array in sorted order.
smallest = p[np.argsort(p)[:n]]  # Returns the indexes that would arrange the array in sorted order.

print("Largest :", largest)  # Displays the value/result labeled "Largest" so it is easy to identify in the output.
print("Smallest:", smallest)  # Displays the value/result labeled "Smallest" so it is easy to identify in the output.

print("\nUsing argpartition:")  # Displays the value/result labeled "Using argpartition" so it is easy to identify in the output.
print("Largest :", p[np.argpartition(-p, n - 1)[:n]])  # Prints the result: partially rearranges indexes so the requested smallest/largest group can be selected efficiently.
print("Smallest:", p[np.argpartition(p, n - 1)[:n]])  # Prints the result: partially rearranges indexes so the requested smallest/largest group can be selected efficiently.
```

## 9.5 Compare Arrays with `np.allclose()`

```python
# ============================================================================
# . NOTES
# - .: 9.5 Compare Arrays with `np.allclose()`.
# - np.arange(start, stop, step) creates regularly spaced values using a STEP SIZE; the stop
#   value is excluded.
# - np.allclose(a, b) checks whether numeric arrays are approximately equal within a small
#   floating-point tolerance.
# - The slice [::-1] uses a step of -1, which reads elements in reverse order.
# ============================================================================

d1 = np.arange(0, 10)  # Creates values from 0 to 9 with the default step of 1; the stop value 10 is excluded.
d2 = np.arange(0, 10)  # Creates values from 0 to 9 with the default step of 1; the stop value 10 is excluded.
d3 = np.arange(10, 20)  # Creates regularly spaced values using a step size; the stop value is excluded.
d4 = d1[::-1]  # Subtracts the values/arrays on the right and stores the result in this variable.

print("d1 vs d2:", np.allclose(d1, d2))  # Prints the result: checks whether corresponding numeric values are equal within a small floating-point tolerance.
print("d1 vs d3:", np.allclose(d1, d3))  # Prints the result: checks whether corresponding numeric values are equal within a small floating-point tolerance.
print("d1 vs d4:", np.allclose(d1, d4))  # Prints the result: checks whether corresponding numeric values are equal within a small floating-point tolerance.
```

## 9.6 Unique Values and Frequency

```python
# ============================================================================
# . NOTES
# - .: 9.6 Unique Values and Frequency.
# - np.array(...) converts a Python sequence (such as a list) into a NumPy ndarray.
# - .argmax() returns the INDEX of the largest value, not the value itself.
# - np.unique(array, return_counts=True) returns distinct values and, optionally, how many times
#   each occurs.
# - return_counts=True asks np.unique to also return how often each unique value occurs.
# ============================================================================

b = np.array([10, 10, 10, 20, 30, 20, 30, 30, 20, 10, 10, 30, 10])  # Creates a NumPy array from the supplied Python values.

values, counts = np.unique(b, return_counts=True)  # Returns unique values and can also return how many times each value occurs.

print("Unique values:", values)  # Displays the value/result labeled "Unique values" so it is easy to identify in the output.
print("Counts       :", counts)  # Displays the value/result labeled "Counts" so it is easy to identify in the output.
print("Most frequent:", values[np.argmax(counts)])  # Prints the result: returns the index of the largest value.
```

## 11.3 Dot Product — Several Equivalent Methods

```python
# ============================================================================
# . NOTES
# - .: 11.3 Dot Product — Several Equivalent Methods.
# - np.array(...) converts a Python sequence (such as a list) into a NumPy ndarray.
# - np.multiply(a, b) performs element-wise multiplication; it is NOT matrix multiplication.
# - np.dot(...) computes a dot product; for 1D vectors it gives a scalar, while for 2D arrays it
#   performs matrix multiplication.
# - np.inner(a, b) computes the inner product over the last dimensions; for 1D vectors it is the
#   usual dot product.
# - np.matmul(a, b) performs matrix multiplication; it follows row-by-column multiplication
#   rules.
# - The @ operator means dot product for 1D vectors and matrix multiplication for 2D matrices.
# - For two vectors, the dot product multiplies matching components and then adds those
#   products, producing one scalar.
# ============================================================================

a1 = np.array([1, 2, 3])  # Creates a NumPy array from the supplied Python values.
a2 = np.array([4, 5, 6])  # Creates a NumPy array from the supplied Python values.

print("Using @          :", a1 @ a2)  # Prints the result of matrix multiplication/dot product performed with the @ operator.
print("Using np.dot     :", np.dot(a1, a2))  # Prints the result: calculates a dot product; for matrices it performs row-by-column matrix multiplication.
print("Using np.inner   :", np.inner(a1, a2))  # Prints the result: calculates the inner product by multiplying matching components and summing the products.
print("Multiply + sum   :", np.multiply(a1, a2).sum())  # Prints the result: multiplies corresponding elements element by element; this is not matrix multiplication.
print("Using np.matmul  :", np.matmul(a1, a2))  # Prints the result: performs matrix multiplication using row-by-column multiplication rules.
```

## 11.4 Length / Magnitude of a Vector

```python
# ============================================================================
# . NOTES
# - .: 11.4 Length / Magnitude of a Vector.
# - np.array(...) converts a Python sequence (such as a list) into a NumPy ndarray.
# - np.sqrt(array) calculates the square root of each element.
# - np.dot(...) computes a dot product; for 1D vectors it gives a scalar, while for 2D arrays it
#   performs matrix multiplication.
# - np.linalg.norm(vector) returns its magnitude/length (Euclidean norm by default).
# ============================================================================

v = np.array([1, 2, 3, 4, 5, 6])  # Creates a NumPy array from the supplied Python values.

length_manual = np.sqrt(np.dot(v, v))  # Calculates the square root of every array element.
length_numpy = np.linalg.norm(v)  # Calculates vector or matrix magnitude; for a vector this gives its Euclidean length by default.

print("Manual:", length_manual)  # Displays the value/result labeled "Manual" so it is easy to identify in the output.
print("NumPy :", length_numpy)  # Displays the value/result labeled "NumPy" so it is easy to identify in the output.
```

## 11.5 Normalize a Vector

```python
# ============================================================================
# . NOTES
# - .: 11.5 Normalize a Vector.
# - np.array(...) converts a Python sequence (such as a list) into a NumPy ndarray.
# - np.linalg.norm(vector) returns its magnitude/length (Euclidean norm by default).
# - Normalization divides a vector by its magnitude so the result has length 1 while keeping the
#   same direction.
# ============================================================================

v = np.array([2.0, 3.0])  # Creates a NumPy array from the supplied Python values.

normalized = v / np.linalg.norm(v)  # Calculates vector or matrix magnitude; for a vector this gives its Euclidean length by default.

print("Original  :", v)  # Displays the value/result labeled "Original" so it is easy to identify in the output.
print("Normalized:", normalized)  # Displays the value/result labeled "Normalized" so it is easy to identify in the output.
print("New length:", np.linalg.norm(normalized))  # Prints the result: calculates vector or matrix magnitude; for a vector this gives its Euclidean length by default.
```

## 14.10 Matrix Dot Product / Matrix Multiplication

```python
# ============================================================================
# . NOTES
# - .: 14.10 Matrix Dot Product / Matrix Multiplication.
# - np.dot(...) computes a dot product; for 1D vectors it gives a scalar, while for 2D arrays it
#   performs matrix multiplication.
# - np.matmul(a, b) performs matrix multiplication; it follows row-by-column multiplication
#   rules.
# - The @ operator means dot product for 1D vectors and matrix multiplication for 2D matrices.
# ============================================================================
M = np.array([[1, 2, 3],  # Creates a NumPy array from the supplied Python values.
              [4, -3, 6],  # Adds one row/list of values to the array being defined.
              [7, 8, 0]])  # Adds the final row/list of values for the array being defined.

N = np.array([[1, 1, 1],  # Creates a NumPy array from the supplied Python values.
              [2, 2, 2],  # Adds one row/list of values to the array being defined.
              [3, 3, 3]])  # Adds the final row/list of values for the array being defined.


print("Using @:\n", M @ N)  # Prints the result of matrix multiplication/dot product performed with the @ operator.
print("\nUsing matmul:\n", np.matmul(M, N))  # Prints the result: performs matrix multiplication using row-by-column multiplication rules.
print("\nUsing dot:\n", np.dot(M, N))  # Prints the result: calculates a dot product; for matrices it performs row-by-column matrix multiplication.
```

## 14.12 Sum of Matrix Elements — Total, Column-Wise, Row-Wise

```python
# ============================================================================
# . NOTES
# - .: 14.12 Sum of Matrix Elements — Total, Column-Wise, Row-Wise.
# - np.sum(...) adds values; with axis=0 it returns one result per column, and with axis=1 one
#   result per row for a 2D array.
# - For a 2D array, axis=0 means work down the rows, so the result contains one value for each
#   COLUMN.
# - For a 2D array, axis=1 means work across the columns, so the result contains one value for
#   each ROW.
# ============================================================================


N = np.array([[1, 1, 1],  # Creates a NumPy array from the supplied Python values.
              [2, 2, 2],  # Adds one row/list of values to the array being defined.
              [3, 3, 3]])  # Adds the final row/list of values for the array being defined.


print("Total sum   :", np.sum(N))  # Prints the result: adds array values; axis controls the direction when an axis is supplied.
print("Column-wise :", np.sum(N, axis=0))  # Prints the result: adds values down the rows and returns one sum for each column because axis=0.
print("Row-wise    :", np.sum(N, axis=1))  # Prints the result: adds values across the columns and returns one sum for each row because axis=1.
```
