## 0. Before We Start — The Main Idea

Pandas mainly gives us two objects:

- **Series** → one-dimensional data, similar to one column.
- **DataFrame** → two-dimensional data, similar to an Excel table.

A DataFrame is made of multiple Series placed together.

## 1. Import Pandas

```python
# Import the pandas library and give it the standard short name pd.
import pandas as pd

# Change the display setting so Pandas shows floating-point numbers with two decimal places.
pd.set_option("display.float_format", "{:.2f}".format)

# Display the installed Pandas version.
print("Pandas version:", pd.__version__)
```

```python
pd.set_option("display.max_rows", 100)  # Shows up to 100 rows in the output.
pd.set_option("display.max_rows", None)  # Shows all rows without limiting them.
pd.set_option("display.max_columns", 50)  # Shows up to 50 columns in the output.
pd.set_option("display.max_columns", None)  # Shows all columns.
pd.set_option("display.float_format", "{:.2f}".format)  # Displays float values with 2 decimal places.
pd.set_option("display.float_format", "{:,.2f}".format)  # Displays floats with commas and 2 decimal places, like 1,234.56.
pd.set_option("display.max_colwidth", 100)  # Shows up to 100 characters inside a column.
pd.set_option("display.max_colwidth", None)  # Shows complete text inside columns without shortening it.
pd.set_option("display.expand_frame_repr", False)  # Tries to display wide DataFrames in one line instead of wrapping.
pd.get_option("display.max_rows")  # Returns the current value of the max_rows option.
pd.describe_option("display.max_rows")  # Shows information about the max_rows option.
pd.describe_option()  # Shows information about all available Pandas options.
pd.reset_option("display.max_rows")  # Resets max_rows back to its default value.
pd.reset_option("display.float_format")  # Resets float formatting back to default.
pd.reset_option("all")  # Resets all Pandas options back to their default values.
```

## 2. Pandas Series

A **Series** stores a one-dimensional collection of values.

Think of it like a single column in a table.

### 2.1 Create a Series from a Python List

```python
# Create a normal Python list containing marks.
marks_list = [78, 85, 92, 66, 88]

# Convert the Python list into a Pandas Series.
marks = pd.Series(marks_list)

# Display the complete Series.
marks
```

### 2.2 Create a Series with Custom Index Labels

```python
# Create a Series and give each value a meaningful index label.
student_marks = pd.Series(
    [78, 85, 92, 66, 88],
    index=["Aman", "Priya", "Rohit", "Neha", "Karan"]
)

# Display the Series.
student_marks
```

### 2.3 Create a Series from a Dictionary

```python
# Create a Python dictionary where keys will become indexes.
city_population = {
    "Delhi": 32,
    "Mumbai": 21,
    "Bengaluru": 14,
    "Chennai": 12
}

# Convert the dictionary into a Pandas Series.
population = pd.Series(city_population)

# Display the Series.
population
```

### 2.4 Important Series Properties

```python
# Display the raw values stored inside the Series.
print("Values:", student_marks.values)
# Display the index labels of the Series.

print("Index:", student_marks.index)
# Display the data type of the Series values.
print("Data type:", student_marks.dtype)

# Display the total number of values.
print("Size:", student_marks.size)

# Display the shape of the Series.
print("Shape:", student_marks.shape)

# Display the number of dimensions.
print("Dimensions:", student_marks.ndim)
```

### 2.5 Access Values from a Series

```python
# Access a value using its label.
print("Priya's marks:", student_marks.loc["Priya"])

# Access a value using its numerical position.
print("First student's marks:", student_marks.iloc[0])

# Select multiple students using labels.
print(student_marks.loc[["Aman", "Rohit", "Karan"]])

# we can go with slicing as well
print(student_marks[0:3])
```

### 2.6 Filter a Series

```python
# Create a Boolean condition that is True where marks are 80 or more.
high_marks_condition = student_marks >= 80

# Display the True/False result of the condition.
print(high_marks_condition)

# Use the Boolean condition to keep only students with marks of 80 or more.
high_scorers = student_marks[high_marks_condition]

# Display the filtered Series.
print(high_scorers)
```

## 3. Pandas DataFrame

A **DataFrame** is a two-dimensional table with rows and columns.

Each column in a DataFrame is itself a Pandas Series.

### 3.1 Create a DataFrame from a Dictionary

```python
# Create a dictionary where each key will become a column.
student_data = {
    "Name": ["Aman", "Priya", "Rohit", "Neha"],
    "Age": [21, 22, 23, 21],
    "City": ["Delhi", "Noida", "Ghaziabad", "Delhi"],
    "Marks": [78, 91, 85, 88]
}
# Convert the dictionary into a DataFrame.
students = pd.DataFrame(student_data)

# Display the DataFrame.
students
```

### 3.2 Basic DataFrame Properties

```python
# Display the number of rows and columns as a tuple.
print("Shape:", students.shape)

# Display only the column names.
print("Columns:", students.columns)

# Display the row index.
print("Index:", students.index)

# Display the data type of every column.
print("Data types:")
print(students.dtypes)

# Display the total number of values in the DataFrame.
print("Total values:", students.size)

# Display the number of dimensions.
print("Dimensions:", students.ndim)
```

### 3.3 One Column vs Multiple Columns

This difference is important:

- `df["Name"]` returns a **Series**
- `df[["Name"]]` returns a **DataFrame**

```python
# Select one column with single square brackets.
one_column_series = students["Name"]

# Select one column but keep the result as a DataFrame.
one_column_dataframe = students[["Name"]]

# Display the type of the first result.
print(type(one_column_series))

# Display the type of the second result.
print(type(one_column_dataframe))
```

## 8. Filtering Data

### 8.1 One Condition

```python
# Create a condition that checks whether Rating is 4.5 or higher.
rating_condition = sales["Rating"] >= 4.5

# Use the condition to keep only matching rows.
high_rated_sales = sales[rating_condition]

# Display selected columns from the filtered data.
high_rated_sales[["Name", "Product Name", "Rating"]]
```

### 8.2 Multiple Conditions with `&`

```python
# Create the first condition for ratings of at least 4.
condition_one = sales["Rating"] >= 4.0

# Create the second condition for sale prices above 200.
condition_two = sales["Sale Price"] > 200

# Combine both conditions using the AND operator &.
filtered_sales = sales[condition_one & condition_two]

# Display the filtered rows.
filtered_sales[["Name", "Sale Price", "Rating"]]
```

### 8.3 OR Condition with `|`

```python
# Create a condition for cash payments.
cash_condition = sales["Payment Mode"] == "Cash"

# Create another condition for UPI payments.
upi_condition = sales["Payment Mode"] == "UPI"

# Keep rows where either condition is True.
cash_or_upi = sales[cash_condition | upi_condition]

# Display a few useful columns.
cash_or_upi[["Name", "Payment Mode", "Sale Price"]]
```

### 8.4 `isin()`

```python
# Create a list of payment methods we want to keep.
wanted_methods = ["UPI", "Card"]

# Check whether each Payment Mode exists in the list.
method_condition = sales["Payment Mode"].isin(wanted_methods)

# Filter the DataFrame using the condition.
sales[method_condition][["Name", "Payment Mode"]]
```

### 8.5 `between()`

```python
# Check which Sale Price values are between 200 and 500, including both limits.
price_condition = sales["Sale Price"].between(200, 500)

# Display matching rows.
sales[price_condition][["Name", "Sale Price"]]
```

## 10. Missing Values

Missing values usually appear as `NaN`.

Before filling or deleting them, first **measure** the problem.

### 10.1 Find Missing Values

```python
# Return True where a value is missing and False where a value exists.
missing_matrix = sales.isna()

# Display the first five rows of the True/False result.
missing_matrix.head()
```

```python
# Count missing values in every column.
missing_per_column = sales.isna().sum()

# Display the count.
missing_per_column
```

```python
# Count all missing values in the complete DataFrame.
total_missing = sales.isna().sum().sum()

# Display the total.
print("Total missing values:", total_missing)
```

### 10.2 Check Missing Values in One Column

```python
# Check which values are missing only in the MRP column.
mrp_missing = sales["MRP"].isna()

# Count how many MRP values are missing.
print("Missing MRP values:", mrp_missing.sum())
```

### 10.3 `dropna()` — Remove Rows or Columns

```python
# Create a fresh copy so the original sales DataFrame is not changed.
drop_demo = sales.copy()

# Remove rows that contain at least one missing value and return a new DataFrame.
rows_without_missing = drop_demo.dropna(axis=0)

# Display the shape before and after removing rows.
print("Before:", drop_demo.shape)
print("After removing rows:", rows_without_missing.shape)
```

```python
# Remove columns that contain at least one missing value and return a new DataFrame.
columns_without_missing = drop_demo.dropna(axis=1)

# Display the remaining column names.
print(columns_without_missing.columns.tolist())
```

### 10.4 Fill Missing Values with a Fixed Value

```python
# Create a fresh copy for the fill demonstration.
fill_demo = sales.copy()

# Fill missing Product Name values with a clear text label.
fill_demo["Product Name"] = fill_demo["Product Name"].fillna("Unknown Product")

# Display the first ten Product Name values.
fill_demo["Product Name"].head(10)
```

### 10.5 Fill Numeric Missing Values with Mean, Median, and Mode

```python
# Create a fresh copy for numeric filling.
numeric_fill_demo = sales.copy()

# Calculate the mean of Sale Price while automatically ignoring missing values.
sale_price_mean = numeric_fill_demo["Sale Price"].mean()

# Fill missing Sale Price values using the mean.
numeric_fill_demo["Sale Price"] = numeric_fill_demo["Sale Price"].fillna(sale_price_mean)

# Display the calculated mean and the remaining missing count.
print("Mean Sale Price:", sale_price_mean)
print("Missing Sale Price after fill:", numeric_fill_demo["Sale Price"].isna().sum())
```

```python
# Create another fresh copy for the median example.
median_fill_demo = sales.copy()

# Calculate the median MRP.
mrp_median = median_fill_demo["MRP"].median()

# Fill missing MRP values with the median.
median_fill_demo["MRP"] = median_fill_demo["MRP"].fillna(mrp_median)

# Display the median and remaining missing count.
print("Median MRP:", mrp_median)
print("Missing MRP after fill:", median_fill_demo["MRP"].isna().sum())
```

```python
# Create another fresh copy for the mode example.
mode_fill_demo = sales.copy()

# Get the most common Rating and take the first mode value.
rating_mode = mode_fill_demo["Rating"].mode()[0]

# Fill missing Rating values with the mode.
mode_fill_demo["Rating"] = mode_fill_demo["Rating"].fillna(rating_mode)

# Display the mode.
print("Mode Rating:", rating_mode)
```

> **Teaching point:** Do not automatically use mean for every numeric column.  
> The correct treatment depends on what the column represents and why the value is missing.

## 11. Data Types and Date Conversion

### 11.1 Convert a Column with `astype()`

```python
# Create a copy for data-type experiments.
type_demo = sales.copy()

# Convert Quantity from integer to float for demonstration.
type_demo["Quantity"] = type_demo["Quantity"].astype(float)

# Display the new data type.
print(type_demo["Quantity"].dtype)
```

### 11.2 Convert Mixed Date Text to Real Datetime

```python
# Create a copy so the original Sale Date text remains available for comparison.
date_demo = sales.copy()

# Convert Sale Date into Pandas datetime while treating day as the first date component.
date_demo["Sale Date"] = pd.to_datetime(
    date_demo["Sale Date"],
    dayfirst=True,
    errors="coerce"
)

# Display the converted date column and its data type.
print(date_demo[["Sale Date"]].head(15))
print(date_demo["Sale Date"].dtype)
```

`errors="coerce"` means values that cannot be converted become `NaT` instead of stopping the program with an error.

`NaT` means **Not a Time**, which is the datetime version of a missing value.

### 11.3 Extract Date Parts

```python
# Extract the year from the datetime column.
date_demo["Year"] = date_demo["Sale Date"].dt.year

# Extract the month number.
date_demo["Month"] = date_demo["Sale Date"].dt.month

# Extract the day of the month.
date_demo["Day"] = date_demo["Sale Date"].dt.day

# Extract the weekday name.
date_demo["Day Name"] = date_demo["Sale Date"].dt.day_name()

# Display the new date-related columns.
date_demo[["Sale Date", "Year", "Month", "Day", "Day Name"]].head(10)
```

## 15. Add, Rename, and Delete Columns

### 15.1 Add Calculated Columns

```python
# Start from data with exact duplicate rows removed.
analysis_sales = sales.drop_duplicates().copy()

# Fill missing Sale Price values with the median so calculations can continue.
analysis_sales["Sale Price"] = analysis_sales["Sale Price"].fillna(
    analysis_sales["Sale Price"].median()
)

# Calculate gross sales before discount.
analysis_sales["Gross Sales"] = (
    analysis_sales["Sale Price"] * analysis_sales["Quantity"]
)

# Calculate the discount amount.
analysis_sales["Discount Amount"] = (
    analysis_sales["Gross Sales"] * analysis_sales["Discount %"] / 100
)

# Calculate net sales after discount.
analysis_sales["Net Sales"] = (
    analysis_sales["Gross Sales"] - analysis_sales["Discount Amount"]
)

# Display the calculated columns.
analysis_sales[
    [
        "Name",
        "Sale Price",
        "Quantity",
        "Discount %",
        "Gross Sales",
        "Discount Amount",
        "Net Sales",
    ]
].head()
```

### 15.2 Rename a Column

```python
# Create a copy so the main analysis DataFrame keeps its original column names.
rename_demo = analysis_sales.copy()

# Rename Sale Price to Selling Price.
rename_demo = rename_demo.rename(
    columns={"Sale Price": "Selling Price"}
)

# Display the new column names.
print(rename_demo.columns.tolist())
```

## 17. `map()`, `replace()`, and `apply()`

### Quick Difference

- `map()` → usually used on a **Series**, often for one-to-one mapping.
- `replace()` → replace known values with other values.
- `apply()` → apply a custom function to a Series or along DataFrame rows/columns.

### 17.1 `map()` with a Dictionary

```python
# Create a dictionary that converts payment modes into broader payment groups.
payment_group_map = {
    "UPI": "Digital",
    "Card": "Digital",
    "Net Banking": "Digital",
    "Cash": "Offline"
}

# Create a new column by mapping each Payment Mode to its group.
analysis_sales["Payment Group"] = analysis_sales["Payment Mode"].map(
    payment_group_map
)

# Display the result.
analysis_sales[["Payment Mode", "Payment Group"]].head(10)
```

### 17.2 `apply()` with a Custom Function

```python
# Define a function that converts numeric ratings into simple text categories.
def rating_category(rating):
    # Return Unknown when the value is missing.
    if pd.isna(rating):
        return "Unknown"
    # Return Excellent for ratings of 4.5 or more.
    if rating >= 4.5:
        return "Excellent"
    # Return Good for ratings of 4.0 or more.
    if rating >= 4.0:
        return "Good"
    # Return Average for all remaining ratings.
    return "Average"

# Apply the function to every value in the Rating column.
analysis_sales["Rating Category"] = analysis_sales["Rating"].apply(
    rating_category
)

# Display the result.
analysis_sales[["Rating", "Rating Category"]].head(10)
```

### 17.3 Row-wise `apply(axis=1)`

```python
# Define a function that reads values from one complete row.
def order_size(row):
    # Return Bulk when Quantity is 4.
    if row["Quantity"] >= 4:
        return "Bulk"
    # Return Regular for smaller quantities.
    return "Regular"

# Apply the function row by row using axis=1.
analysis_sales["Order Size"] = analysis_sales.apply(
    order_size,
    axis=1
)

# Display the result.
analysis_sales[["Quantity", "Order Size"]].head(10)
```

## 20. Concatenating DataFrames with `pd.concat()`

Use `concat()` when tables have the same or compatible structure and you want to stack them.

- `axis=0` → stack rows vertically
- `axis=1` → place columns side by side

### 20.1 Concatenate Rows

```python
# Create the first small DataFrame.
class_a = pd.DataFrame(
    {
        "Name": ["Aman", "Priya"],
        "Marks": [80, 92]
    }
)

# Create the second small DataFrame.
class_b = pd.DataFrame(
    {
        "Name": ["Rohit", "Neha"],
        "Marks": [85, 88]
    }
)

# Stack the two DataFrames vertically and create a fresh continuous index.
combined_students = pd.concat(
    [class_a, class_b],
    axis=0,
    ignore_index=True
)

# Display the combined rows.
combined_students
```

## 21. Merge Two CSV Files — Real Join Example

Now we will work with multiple files, just like a real project.

Files:

- `sales_data_class.csv` → transaction data
- `product_details.csv` → product master data
- `customer_details.csv` → customer master data

`merge()` is similar to SQL joins.

### 21.1 Load the Product Master File

```python
# Read the product master CSV file.
products = pd.read_csv("product_details.csv")

# Display the first five rows.
products.head()
```

### 21.2 Left Join Sales with Product Details

```python
# Merge every sales row with matching product information.
sales_with_products = pd.merge(
    sales,
    products,
    on="Product Name",
    how="left"
)

# Display useful columns after the merge.
sales_with_products[
    ["Name", "Product Name", "Category", "Brand", "Sale Price"]
].head(12)
```

**Why is `left` useful here?**

We normally want to keep every transaction from the sales file even when a product lookup is missing.

### 21.3 Compare Inner, Left, Right, and Outer Joins

```python
# Keep only keys that exist in both DataFrames.
inner_join = pd.merge(
    sales,
    products,
    on="Product Name",
    how="inner"
)

# Keep every row from the left sales DataFrame.
left_join = pd.merge(
    sales,
    products,
    on="Product Name",
    how="left"
)

# Keep every row from the right products DataFrame.
right_join = pd.merge(
    sales,
    products,
    on="Product Name",
    how="right"
)

# Keep every key from both DataFrames.
outer_join = pd.merge(
    sales,
    products,
    on="Product Name",
    how="outer"
)

# Compare the number of rows created by each join type.
print("Inner join rows:", len(inner_join))
print("Left join rows:", len(left_join))
print("Right join rows:", len(right_join))
print("Outer join rows:", len(outer_join))
```

### 21.4 Use `indicator=True` to See Where Rows Came From

```python
# Perform an outer join and ask Pandas to show the source of each row.
join_check = pd.merge(
    sales,
    products,
    on="Product Name",
    how="outer",
    indicator=True
)

# Count rows that matched, came only from the left, or came only from the right.
join_check["_merge"].value_counts()
```

### 21.5 Merge Customer Details

```python
# Read the customer master CSV file.
customers = pd.read_csv("customer_details.csv")

# Merge customer information into the sales-plus-products DataFrame.
full_sales = pd.merge(
    sales_with_products,
    customers[["Customer ID", "City", "State", "Segment"]],
    on="Customer ID",
    how="left"
)

# Display selected columns from the fully combined data.
full_sales[
    [
        "Customer ID",
        "Name",
        "City",
        "Segment",
        "Product Name",
        "Category",
        "Sale Price",
    ]
].head(12)
```

## 22. `join()` Using Indexes

`join()` is commonly used when the matching key has been placed in the index.

For beginners, remember:

- `merge()` → very flexible, usually joins using columns.
- `join()` → especially convenient when joining by index.

```python
# Put Customer ID in the index of the sales DataFrame.
sales_indexed = sales.set_index("Customer ID")

# Put Customer ID in the index of the customer master DataFrame.
customers_indexed = customers.set_index("Customer ID")

# Join selected customer columns into the sales data by matching indexes.
joined_by_index = sales_indexed.join(
    customers_indexed[["City", "State", "Segment"]],
    how="left"
)

# Convert Customer ID from the index back into a normal column.
joined_by_index = joined_by_index.reset_index()

# Display selected columns.
joined_by_index[
    ["Customer ID", "Name", "City", "State", "Segment"]
].head(10)
```

## 23. Analysis After Merge

### 23.1 Total Sales by Category

```python
# Start from the fully merged table and remove exact duplicate rows.
merged_analysis = full_sales.drop_duplicates().copy()

# Fill missing Sale Price values with the median.
merged_analysis["Sale Price"] = merged_analysis["Sale Price"].fillna(
    merged_analysis["Sale Price"].median()
)

# Calculate Gross Sales.
merged_analysis["Gross Sales"] = (
    merged_analysis["Sale Price"] * merged_analysis["Quantity"]
)

# Calculate Net Sales after discount.
merged_analysis["Net Sales"] = (
    merged_analysis["Gross Sales"]
    * (1 - merged_analysis["Discount %"] / 100)
)

# Group by Category and calculate total Net Sales.
category_sales = (
    merged_analysis
    .groupby("Category", dropna=False)["Net Sales"]
    .sum()
    .sort_values(ascending=False)
)

# Display the result.
category_sales
```

### 23.2 Total Sales by City

```python
# Group by City and calculate total Net Sales.
city_sales = (
    merged_analysis
    .groupby("City", dropna=False)["Net Sales"]
    .sum()
    .sort_values(ascending=False)
)

# Display the first ten cities.
city_sales.head(10)
```

## 24. Correlation — A Gentle Introduction

Correlation measures how strongly numeric columns move together.

- close to `1` → strong positive relationship
- close to `-1` → strong negative relationship
- close to `0` → little linear relationship

Correlation does **not** automatically mean one variable causes another.

```python
# Select numeric columns that are useful for a simple correlation example.
numeric_columns = merged_analysis[
    ["Sale Price", "MRP", "Rating", "Quantity", "Discount %", "Net Sales"]
]

# Calculate the correlation matrix.
correlation_matrix = numeric_columns.corr()

# Display the matrix.
correlation_matrix
```
